```python
genres = df["listed_in"].str.split(",", expand=True)
```

```python
genre_list = []

for i in genres.columns:
    genre_list.extend(genres[i].dropna().str.strip())

genre_series = pd.Series(genre_list)
```

```python
popular_genres = genre_series.value_counts().head(10)

print(popular_genres)
```

```python
plt.bar(popular_genres.index, popular_genres.values)
plt.title("Top 10 Popular Genres")
plt.xlabel("Genres")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
year_trend =df["release_year"].value_counts().sort_index()
```

```python
plt.figure(figsize=(12,5))

plt.plot(year_trend.index, year_trend.values)

plt.title("Netflix Content Release Trend")
plt.xlabel("Release Year")
plt.ylabel("Number of Shows")

plt.grid()

plt.show()
```

```python
rating_dist = df["rating"].value_counts()
```

```python
rating_dist.plot(kind="pie", autopct="%1.1f%%")

plt.title("Rating Distribution")

plt.ylabel("")

plt.show()
```

```python
df["type"].value_counts().plot(kind="bar")

plt.title("Movies vs TV Shows")

plt.show()
```

```python
df["country"].value_counts().head(10).plot(kind="bar")

plt.title("Top Countries Producing Content")

plt.show()
```
